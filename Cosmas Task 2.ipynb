```python
# Part 1: Check if a number is positive, negative, or zero
Testing_number = {-3,-4,-6,0,23,25,30}

def check_number(number):
    if number > 0:
        return "positive"
    elif number < 0:
        return "negative"
    else:
        return "zero"
    
   
```

```python
print("Check Numbers:")
print("-"*40)

print(f"check number(-3): {check_number(-3)}")

print("-" * 30)

print(f"check number(0): {check_number(0)}")

```

```python
# Overal testing Numbers 
# Testing_number = {-3,-4,-6,0,23,25,30}

print("Testing the numbers with logic expression:")
print("-"*40)

print(f"{'Number':<10} {'|'} {'Classification':<15}")
print("-"*40)

for number in Testing_number :
    result = check_number(number)
    print(f"Number {number}   {"|"}  {result}")
    
print("-"*40)
```

```python
# Part 2: Function to calculate sum and product of two integers

def calculate_sum_and_product(a, b):
    """
    Accept two integers and return their sum and product.
    
    Args:
        a (int): First integer
        b (int): Second integer
    
    Returns:
        tuple: (sum, product) of the two integers
    """
    

    sum_result = a + b
    product_result = a * b
    return sum_result, product_result
```

```python
print ("calculate sum and product of two integers:")
print("--"* 30)

calculate_sum_and_product(24, 30)


```

```python
# Testing Part 2: Sum and product calculation
print(" Sum and Product Calculator:")
print ("-" *30)

numbers  = [
    (5, 10),
    (-20, 7),
    (0, 10),
    (-6, -8),
    (12, 0),
    (-15, 3)
]

for a, b in numbers:
    sum_result, product_result = calculate_sum_and_product (a, b)
    print(f"Test Number: {a} and {b}")                #showing the numbers 
    print(f" Sum:  {sum_result}")                #shows the Sum of the numbers
    print(f"  Product: {product_result}")         #retun the products 
    
    print ("-" *30) #separator
    
    print()
    
print ("successfull Code")
```
