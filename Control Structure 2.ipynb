```python
Names = ['Joy','Peter','Joan','Jayden','Annah','Amos','Kelly']
for name in Names:
    print(name)
```

```python
#break = stops the loop pre-maturity 
# break the chain from the list 
for name in Names :
    if name == "Jayden":
        break
    print(name)
        
```

```python
# == sign is a comperator 
for name in Names:
    if name == 'Jayden':
        continue
    print(name)
```
