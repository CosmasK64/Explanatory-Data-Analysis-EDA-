```python
price1= df["Price"].to_numpy()
quantity1 = df["Quantity"].to_numpy()
```

```python
revenue = price1 * quantity1
```

```python
#Adding new column into data frame called Revenue 
df["Revenue"]=revenue
```

```python
df.head()
```

```python
#revenue by category 
summarized= df.groupby('Category')['Revenue'].sum()
print (summarized)
    
```

```python
summarized= df.groupby('Category')['Revenue'].sum()

totalsales=df['Revenue'].sum()

for Category, Revenue in summarized.items():
    percentage=  (Revenue /totalsales) * 100
    print(f"{Category}: ${Revenue} ({percentage.round(2)}%)")
```

```python
Top = df.groupby('Product')['Revenue'].sum().idxmax()
```

```python
print(f'Top Product:{Top}')
```
