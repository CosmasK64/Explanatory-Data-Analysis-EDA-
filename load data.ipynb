```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data (assuming your CSV is already loaded as 'df')
# df = pd.read_csv(r'C:\Users\user\Downloads\states.csv')

# Sample data based on what's visible in your screenshot
data = {
    'State': ['Alabama', 'Alaska', 'Arizona', 'Arkansas', 'California', 'Colorado'],
    'Average Monthly Tax Credit (2016)': [310, 750, 230, 306, 309, 318]
}

# Create DataFrame (or use your existing df)
df_sample = pd.DataFrame(data)

# Method 1: Using matplotlib
plt.figure(figsize=(12, 8))
df_sorted = df_sample.sort_values('Average Monthly Tax Credit (2016)', ascending=False)
bars = plt.bar(df_sorted['State'], df_sorted['Average Monthly Tax Credit (2016)'], 
               color='steelblue', edgecolor='navy', linewidth=1)

plt.title('Average Monthly Tax Credit 2016 by State', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('State', fontsize=12)
plt.ylabel('Average Monthly Tax Credit ($)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)

# Add value labels on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 5,
             f'${int(height)}', ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()

# Method 2: Using seaborn (alternative approach)
plt.figure(figsize=(12, 8))
sns.set_style("whitegrid")
ax = sns.barplot(data=df_sorted, x='State', y='Average Monthly Tax Credit (2016)', 
                 palette='Blues_d')

plt.title('Average Monthly Tax Credit 2016 by State', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('State', fontsize=12)
plt.ylabel('Average Monthly Tax Credit ($)', fontsize=12)
plt.xticks(rotation=45, ha='right')

# Add value labels
for i, v in enumerate(df_sorted['Average Monthly Tax Credit (2016)']):
    ax.text(i, v + 5, f'${int(v)}', ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()

# If you want to use your complete dataset:
# Replace the sample data section with:
# df_sorted = df.sort_values('Average Monthly Tax Credit (2016)', ascending=False)
# 
# Then create the plot with all states:
# plt.figure(figsize=(20, 10))  # Larger figure for all states
# bars = plt.bar(df_sorted['State'], df_sorted['Average Monthly Tax Credit (2016)'], 
#                color='steelblue', edgecolor='navy', linewidth=1)
# 
# plt.title('Average Monthly Tax Credit 2016 by State', fontsize=16, fontweight='bold', pad=20)
# plt.xlabel('State', fontsize=12)
# plt.ylabel('Average Monthly Tax Credit ($)', fontsize=12)
# plt.xticks(rotation=90, ha='center')  # Vertical labels for all states
# plt.grid(axis='y', alpha=0.3)
# 
# # Add value labels (optional for all states - might be crowded)
# for bar in bars:
#     height = bar.get_height()
#     plt.text(bar.get_x() + bar.get_width()/2., height + 5,
#              f'${int(height)}', ha='center', va='bottom', fontsize=8)
# 
# plt.tight_layout()
# plt.show()

# Print summary statistics
print("Summary Statistics for Average Monthly Tax Credit (2016):")
print(f"Mean: ${df_sample['Average Monthly Tax Credit (2016)'].mean():.2f}")
print(f"Median: ${df_sample['Average Monthly Tax Credit (2016)'].median():.2f}")
print(f"Min: ${df_sample['Average Monthly Tax Credit (2016)'].min():.2f}")
print(f"Max: ${df_sample['Average Monthly Tax Credit (2016)'].max():.2f}")
print(f"Standard Deviation: ${df_sample['Average Monthly Tax Credit (2016)'].std():.2f}")
```
