```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
salesdata1 = pd.read_csv(r'C:\Users\user\Downloads\Sales Data.csv')
```

```python
sales= pd.DataFrame(salesdata1)
```

```python
price= np.array(sales ["Price"])
quantity=np.array(sales ["Quantity"])
```

```python
print(f"Price Array Shape: {price.shape}")
print(f"Price Array (first 5): {price[:5]}")
print(f"Quantity Array Shape: {quantity.shape}")
print(f"quantity Array (first 5): {quantity[:5]}")
```

```python
# Multiply price per quantity in the dataframe 
Total_price = np.multiply(price, quantity)
```

```python
print(f"Price Array Shape: {Total_price.shape}")
print(f"Price Array (first 5): {Total_price[:5]}")
```

```python
#dding to the data frame the calculated filed 
sales["Total_price"] = Total_price
```

```python
Total_price_category = sales.groupby('Category')['Total_price'].sum().sort_values(ascending=False)
total_Price = sales['Total_price'].sum()

print("Totalprice Summary by Category:")
for category, Total_price in Total_price_category.items():
    percentage = (Total_price / total_Price) * 100
    print(f"{category:12}: ${Total_price:10,.2f} ({percentage:5.1f}%)")

print(f"\n Total price: ${total_Price:,.2f}")
```

```python
# 5. Find the Top-Performing Product

top_product= np.argmax(Total_price)  # Using NumPy to find max index
top_product2 = sales.iloc[top_product]

print(f"🏆 TOP-PERFORMING PRODUCT:")
print(f"Product:     {top_product2['Product']}")
print(f"Total_price: ${top_product2['Total_price']:,.2f}")
print(f"Price:       ${top_product2['Price']}")
print(f"Quantity:    {top_product2['Quantity']:,}")
print(f"Category:    {top_product2['Category']}")
print(f"Region:      {top_product2['Region']}")
```
