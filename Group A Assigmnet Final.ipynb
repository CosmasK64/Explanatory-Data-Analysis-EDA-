```python
#D, Create a boxplot showing Loan Amount by Credit History.

#plotting boxplot Graph
plt.figure(figsize=(8, 6))
sns.boxplot(x=Bankdata['Credit_History'], y=Bankdata['Loan_Amount'], palette=["#29ea46", "#cc3a3a"])

plt.title('Loan Amount by Credit History', fontsize=16, fontweight='bold')
plt.xlabel('Credit History', fontsize=12, fontweight='bold')
plt.ylabel('Loan Amount', fontsize=12, fontweight='bold')


plt.tight_layout()
plt.show()

# Is there a difference in loan amounts between those with/without credit history?
#finding mean and medium 
for credit in ['Yes', 'No']:
    subset = Bankdata[Bankdata['Credit_History'] == credit]['Loan_Amount']
    print(f"   Credit History {credit}:")
    print(f"     Mean: ${subset.mean():.0f}")
    print(f"     Median: ${subset.median():.0f}")
    print(f"     Count: {len(subset)}")

```

```python
Q1 = Bankdata['Applicant_Income'].quantile(0.25)
Q3 = Bankdata['Applicant_Income'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = Bankdata[(Bankdata['Applicant_IncomeApplicant_Income'] < lower_bound) | (Bankdata['Applicant_Income'] > upper_bound)]

print("Number of outliers detected:", len(outliers))
print(outliers.head())
```

```python
from  scipy import stats

Bankdata['Applicant_Income_Z'] = stats.zscore(Bankdata['Applicant_Income'])

outliers = Bankdata[(Bankdata['Applicant_Income_Z'] > 3) | (Bankdata['Applicant_Income_Z'] < -3)]

print("Number of outliers detected:", len(outliers))
print(outliers[['Applicant_Income', 'Applicant_Income_Z']].head())
```
