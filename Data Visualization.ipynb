```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
data = pd.DataFrame({"product": ["A","B","C"],
                     "Price": [265, 300, 400]})
```

```python
data.head() 
```

```python
plt.plot(data["Price"]) # plot the line graph 
plt.show()
```

```python
sns.lineplot(x = data["product"], y = data["Price"])
plt.show()
```

```python
sns.lineplot(data["Price"])
```

```python
sns.barplot(x= data["product"], y=data["Price"])
plt.show()
```

```python
sns.barplot(x = "product", y = "Price", data=data)
```

```python
sns.barplot(x = "Price", y = "product", data=data)
plt.title("Price per Products")
plt.show()
```

```python
plt.bar(data["product"], data["Price"])
```

```python
plt.bar(data["product"], data["Price"], color="Green")
plt.title("Product per Price")
plt.xlabel("product")
plt.ylabel("Price")
plt.show()
```

```python
#sub plots 
## Row ,column, Chart number

plt.subplot(1,2,1)
plt.plot(data['Price'])

plt.subplot(1,2,2)
sns.barplot(x= data['product'],y=data['Price'])
plt.show()
```

```python
#sub plots 
## Row ,column, Chart number

plt.subplot(2,1,1)
plt.plot(data['Price'])

plt.subplot(2,1,2)
sns.barplot(x= data['product'],y=data['Price'])
plt.show()
```

```python
sns.countplot(x=data['product'])
plt.show
```

```python
# the upper line represent the Q3 
#middles line represent the Mediun or Q2
# the lower line represent the lower quarter Q1

# lower whisker represent the minimum value in the data set 
# upper whisker represent the maximum value of the data set 
#

sns.boxplot(data['Price'])
plt.show()
```
