```python
# Using a dictionary to store and display prices of 5 grocery items
print("display prices of 5 grocery items :") # title
print("-" * 40) # page break
grocery_prices = {"Rice": 2.50,"Beans": 2.0, "Milk": 3.50, "Bread": 4.50,"Eggs": 5.00}

print(grocery_prices)
```

```python
# Script that iterates over the dictionary and prints each key-value pair
print("Iterating over grocery prices:") # title
print("-" * 30) # page break 

for item, price in grocery_prices.items():
     print(f"{item}: ${price:.2f}") # 2 means two decimal places --F-means store as float--f --means a function 
```
