```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Set up the plotting style
plt.style.use('default')
sns.set_palette("husl")
```

```python
salesdata = pd.read_csv(r'C:\Users\user\Downloads\Sales Data.csv')
```

```python
# Create DataFrame
Sales = pd.DataFrame(salesdata)
```

```python
print("=" * 60)
print("SALES DATA ANALYSIS USING NUMPY AND PANDAS")
print("=" * 60)

# 1. Convert relevant columns to NumPy arrays
print("1. CONVERTING RELEVANT COLUMNS TO NUMPY ARRAYS")
print("-" * 50)

price_array = np.array(Sales['Price'])
quantity_array = np.array(Sales['Quantity'])

print(f"Price Array Shape: {price_array.shape}")
print(f"Price Array (first 5): {price_array[:5]}")
print(f"Quantity Array Shape: {quantity_array.shape}")
print(f"Quantity Array (first 5): {quantity_array[:5]}")
```

```python
# 2. Calculate revenue using NumPy
print("2. CALCULATING REVENUE USING NUMPY")
print("-" * 50)

# Element-wise multiplication using NumPy
revenue_array = np.multiply(price_array, quantity_array)

print(f"Revenue Array Shape: {revenue_array.shape}")
print(f"Revenue Array (first 5): {revenue_array[:5]}")
print(f"Total Revenue (using np.sum): ${np.sum(revenue_array):,.2f}")
```

```python
# 2. Calculate revenue using NumPy
print("\n2. CALCULATING REVENUE USING NUMPY")
print("-" * 50)

# Element-wise multiplication using NumPy
revenue_array = np.multiply(price_array, quantity_array)

print(f"Revenue Array Shape: {revenue_array.shape}")
print(f"Revenue Array (first 5): {revenue_array[:5]}")
print(f"Total Revenue (using np.sum): ${np.sum(revenue_array):,.2f}")
```

```python
# 3. Add the calculated revenue back to the DataFrame
print("3. ADDING CALCULATED REVENUE BACK TO DATAFRAME")
print("-" * 50)

Sales['Revenue'] = revenue_array

# Display the enhanced DataFrame
print(f"DataFrame shape after adding Revenue: {Sales.shape}")
print("\n Enhanced DataFrame (sorted by Revenue, descending):")
Sales_sorted = Sales.sort_values('Revenue', ascending=False)
print(Sales_sorted.to_string(index=False))
```

```python
# 4. Summarize Total Revenue by Category
print("\n4. TOTAL REVENUE BY CATEGORY")
print("-" * 50)

revenue_by_category = Sales.groupby('Category')['Revenue'].sum().sort_values(ascending=False)
total_revenue = Sales['Revenue'].sum()

print("Revenue Summary by Category:")
for category, revenue in revenue_by_category.items():
    percentage = (revenue / total_revenue) * 100
    print(f"{category:12}: ${revenue:10,.2f} ({percentage:5.1f}%)")

print(f"\nTotal Revenue: ${total_revenue:,.2f}")
```

```python
# 5. Find the Top-Performing Product
print("\n5. TOP-PERFORMING PRODUCT")
print("-" * 50)

top_product_idx = np.argmax(revenue_array)  # Using NumPy to find max index
top_product = Sales.iloc[top_product_idx]

print(f"🏆 TOP-PERFORMING PRODUCT:")
print(f"Product:    {top_product['Product']}")
print(f"Revenue:    ${top_product['Revenue']:,.2f}")
print(f"Price:      ${top_product['Price']}")
print(f"Quantity:   {top_product['Quantity']:,}")
print(f"Category:   {top_product['Category']}")
print(f"Region:     {top_product['Region']}")
```

```python
# Additional NumPy-based statistics
print("\n6. ADDITIONAL NUMPY STATISTICS")
print("-" * 50)

print(f"Mean Revenue per Product: ${np.mean(revenue_array):,.2f}")
print(f"Median Revenue per Product: ${np.median(revenue_array):,.2f}")
print(f"Standard Deviation: ${np.std(revenue_array):,.2f}")
print(f"Minimum Revenue: ${np.min(revenue_array):,.2f}")
print(f"Maximum Revenue: ${np.max(revenue_array):,.2f}")
```

```python
# Top 5 products by revenue
print(f"\nTop 5 Products by Revenue:")
top_5_indices = np.argsort(revenue_array)[-5:][::-1]  # Get indices of top 5
for i, idx in enumerate(top_5_indices, 1):
    product_name = Sales.iloc[idx]['Product']
    revenue = revenue_array[idx]
    print(f"{i}. {product_name}: ${revenue:,.2f}")
```

```python
# Category-wise analysis using NumPy
print("\n7. CATEGORY-WISE ANALYSIS USING NUMPY")
print("-" * 50)

categories = Sales['Category'].unique()
for category in categories:
    category_mask = Sales['Category'] == category
    category_revenues = revenue_array[category_mask]
    category_quantities = quantity_array[category_mask]
    
    print(f"\n{category}:")
    print(f"  Total Revenue: ${np.sum(category_revenues):,.2f}")
    print(f"  Average Revenue per Product: ${np.mean(category_revenues):,.2f}")
    print(f"  Total Quantity Sold: {np.sum(category_quantities):,}")
    print(f"  Number of Products: {len(category_revenues)}")
```
